# Topic 2: Machine Learning (ML)

## 1. Definition
Machine Learning (ML) is a subset of Artificial Intelligence that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed.

## 2. Why do we need it?
Traditional programming fails when writing explicit rules becomes impossible.

### Example: Spam Detection
**Traditional Programming:** Requires manual 'If-Else' rules (e.g., *If email contains "Win Money" → Spam*). This fails as spammers change tactics.

**Machine Learning:** An algorithm looks at thousands of emails, learns the patterns automatically, and predicts if new mail is spam without manual rules.

## 3. Formula
Machine Learning tries to learn a function:

$$Y = f(X)$$

* **X**: Input Features (e.g., Area, Bedrooms)
* **Y**: Output (e.g., Price)
* **f**: Model learned from data

## 4. Example
**House Price Prediction**
* **Input:** Area = 1200 sq.ft, Bedrooms = 2, Age = 5 years
* **Output:** ₹75 Lakhs

## 5. Working
1. **Collect Data**
2. **Train Model (Learning Algorithm)**
3. **Learn Patterns**
4. **Predict New Data**

## 6. Where is it used?
* Email Spam & Fraud Detection
* Face Recognition
* Netflix Recommendations
* Medical Diagnosis

## 7. Advantages & 8. Limitations
| Advantages | Limitations |
| :--- | :--- |
| Learns automatically | Needs high-quality data |
| Improves with more data | Can 'overfit' (memorize instead of learn) |
| Reduces manual rule writing | Requires significant computation |

## 9. Python Example
Using `scikit-learn` to demonstrate the `fit()` and `predict()` concept.

````python
from sklearn.linear_model import LinearRegression
import numpy as np

# 1. Prepare Data (X = Area in sq ft, y = Price in Lakhs)
X = np.array([[1000], [1500], [2000], [2500]])
y = np.array([50, 75, 100, 125])

# 2. Initialize the Model
model = LinearRegression()

# 3. Fit (The model learns the relationship)
model.fit(X, y)

# 4. Predict (Predicting price for a 1200 sq ft house)
new_house = np.array([[1200]])
prediction = model.predict(new_house)

print(f"Predicted price for a 1200 sq.ft house: ₹{prediction[0]:.2f} Lakhs")
````

### Visualizing ($Y = f(X)$)
This graph shows the data points (actual prices) and the line (the model) that Machine Learning has learned to fit.

````python
import matplotlib.pyplot as plt

# Generate points for the regression line
X_line = np.linspace(800, 2700, 100).reshape(-1, 1)
y_line = model.predict(X_line)

plt.figure(figsize=(10, 6))
# Plot original data points
plt.scatter(X, y, color='blue', label='Actual Data (Training)')
# Plot the prediction line
plt.plot(X_line, y_line, color='red', linewidth=2, label='ML Model: f(x)')
# Highlight the new prediction
plt.scatter(new_house, prediction, color='green', marker='x', s=100, label=f'Prediction for {new_house[0][0]} sq.ft')

plt.title('House Price Prediction: $Y = f(X)$')
plt.xlabel('Area (sq.ft) [X]')
plt.ylabel('Price (Lakhs) [Y]')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
````

# Topic 3: Deep Learning (DL)

## 1. Definition
Deep Learning (DL) is a subset of Machine Learning that uses Artificial Neural Networks (ANNs) with multiple hidden layers to automatically learn complex patterns from data.

## 2. Why do we need it?
Traditional ML struggles with very large and unstructured data like images, videos, and speech. Deep Learning learns features automatically instead of requiring manual feature engineering.

### Comparison:
* **Machine Learning:** Image → Human extracts features → Algorithm → Output
* **Deep Learning:** Image → Neural Network (Automatic extraction) → Output

## 3. Formula
A single neuron computes a weighted sum followed by an activation function:

$$z = \sum (w_i x_i) + b$$
$$Output = Activation(z)$$

## 4. Example: Image Classification
The network automatically learns patterns in layers:
**Pixels** → **Edges** → **Shapes** → **Eyes** → **Face** → **Cat**

## 5. Working
1. **Forward Pass:** Input travels through layers to make a prediction.
2. **Loss Calculation:** Compare prediction to actual result.
3. **Backpropagation:** Calculate error gradients.
4. **Update Weights:** Adjust connections to reduce error.

## 6. Where is it used?
* ChatGPT, Gemini, Claude
* Face & Speech Recognition
* Self-driving Cars
* Image Generation (DALL-E, Midjourney)

## 7. Advantages & 8. Limitations
| Advantages | Limitations |
| :--- | :--- |
| Automatic feature extraction | Requires massive datasets |
| High accuracy on complex data | Computationally expensive (Requires GPUs) |
| Scales with more data | 'Black Box' (Hard to interpret) |

## 9. Python Example
Using `TensorFlow/Keras` to define a simple neural network architecture.

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

# Define a simple architecture using the Input layer (best practice)
model = Sequential([
    Input(shape=(10,)),
    # Hidden layer with 64 neurons and ReLU activation
    Dense(64, activation="relu"),
    # Output layer with Sigmoid activation for binary classification
    Dense(1, activation="sigmoid")
])

model.summary()
````

# Topic 6: Supervised Learning

## 1. Definition
Supervised Learning is a type of Machine Learning where the model learns from labeled data, i.e., both the input (**X**) and the correct output (**Y**) are provided during training. The goal is to learn the relationship $Y = f(X)$ to predict outcomes for unseen data.

## 2. Important Terminology
*   **Feature (X):** The input variables used for prediction (e.g., Hours Studied, Attendance).
*   **Label / Target (Y):** The correct answer we want the model to learn (e.g., Pass/Fail, Price).
*   **Training Data:** The subset of data used for the model to learn from.
*   **Testing Data:** New, unseen data used to evaluate how well the model has learned.

## 3. Types of Supervised Learning

| Type | Output Type | Examples |
| :--- | :--- | :--- |
| **Classification** | Categorical (Classes) | Spam vs. Not Spam, Pass vs. Fail, Cat vs. Dog |
| **Regression** | Continuous (Numbers) | House Price, Temperature, Stock Price |

## 4. Workflow
**Training Data (X, Y)** $\rightarrow$ **Learning Algorithm** $\rightarrow$ **Trained Model** $\rightarrow$ **New X** $\rightarrow$ **Prediction Y**

## 5. Python Example: Classification with Decision Trees
We will use a `DecisionTreeClassifier` to predict if a student with **6 hours of study and 80% attendance** will pass.

````python
from sklearn.tree import DecisionTreeClassifier
import pandas as pd

# 1. Prepare Training Data
df_supervised = pd.DataFrame({
    "Hours": [2, 4, 5, 8],
    "Attendance": [60, 70, 75, 95],
    "Pass": [0, 0, 1, 1]  # 0 = No, 1 = Yes
})

X = df_supervised[["Hours", "Attendance"]]
y = df_supervised["Pass"]

# 2. Initialize and Train Model
clf = DecisionTreeClassifier()
clf.fit(X, y)

# 3. Predict for a new student (using a DataFrame to include feature names)
new_student = pd.DataFrame([[6, 80]], columns=["Hours", "Attendance"])
prediction = clf.predict(new_student)
result = "Pass" if prediction[0] == 1 else "Fail"

print(f"Prediction for 6 hours study & 80% attendance: {result} (Class {prediction[0]})")
````

➡ Supervised learning is highly accurate but depends on expensive labeled data. This limitation leads us to **Unsupervised Learning**, where labels are unavailable.

# Topic 7: Unsupervised Learning

## 1. Definition
Unsupervised Learning is a type of Machine Learning where the model learns only from input data (**X**) without any labels (**Y**). The goal is to discover hidden patterns, groups, or relationships in the data.

## 2. Why do we need it?
In many real-world problems, labels are unavailable. For example, a mall has customer data but doesn't know the categories. The ML model automatically groups similar customers based on behavior.

## 3. Workflow & Visual Representation
There is no target variable ($Y$), only the features ($X$).

```text
Unlabeled Data (X)
       ↓
Learning Algorithm (e.g., K-Means)
       ↓
Find Similarities (Distance Metrics)
       ↓
Create Clusters / Insights
```

## 4. Types of Unsupervised Learning
| Type | Description | Common Algorithms |
| :--- | :--- | :--- |
| **Clustering** | Groups similar data points | K-Means, Hierarchical, DBSCAN |
| **Dimensionality Reduction** | Reduces features while keeping info | PCA, t-SNE, UMAP |
| **Association** | Finds relationships between items | Apriori, Eclat |
| **Anomaly Detection** | Finds unusual observations | Isolation Forest, One-Class SVM |

## 5. Python Example: Customer Segmentation (K-Means)
We will use `KMeans` to automatically group customers based on their **Age** and **Income** without providing any labels.

````python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# 1. Create the unlabeled dataset
df_unsupervised = pd.DataFrame({
    "Age": [22, 21, 45, 48, 23, 44, 50, 20],
    "Income": [25, 20, 90, 85, 28, 80, 95, 22]
})

# 2. Initialize and Train Model
# We tell the model to find 2 clusters (e.g., Young/Low-Income vs. Older/High-Income)
model = KMeans(n_clusters=2, random_state=42, n_init='auto')
df_unsupervised["Cluster"] = model.fit_predict(df_unsupervised)

# 3. Display Results
print("Data with Automatically Generated Cluster Labels:")
display(df_unsupervised)

# 4. Visualize the clusters
plt.figure(figsize=(8, 5))
plt.scatter(df_unsupervised['Age'], df_unsupervised['Income'], c=df_unsupervised['Cluster'], cmap='viridis', s=100)
plt.title('Unsupervised Learning: K-Means Clustering')
plt.xlabel('Age')
plt.ylabel('Income (k$)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
````

## 6. Advantages & Limitations
| Advantages | Limitations |
| :--- | :--- |
| No expensive labels required | Hard to evaluate accuracy |
| Discovers hidden patterns | Results can be sensitive to parameters |
| Great for exploratory analysis | Cluster meanings aren't always obvious |

➡ Sometimes labels are expensive, but we have a *few* labeled samples. This middle ground is known as **Semi-Supervised Learning**.

# Topic 8: Semi-Supervised Learning

## 1. Definition
Semi-Supervised Learning is a hybrid type of Machine Learning where the model is trained using a **small amount of labeled data** combined with a **large amount of unlabeled data**.

## 2. Why do we need it?
Labeling data is often expensive and requires human experts. For example, in **Medical Imaging**, a doctor might only have time to label 500 out of 10,000 X-ray images. Semi-supervised learning ensures the remaining 9,500 images are not wasted.

## 3. Workflow & Visual Representation
Training Data = $(X, Y)_{labeled} + (X)_{unlabeled}$

```text
Small Labeled Dataset + Large Unlabeled Dataset
            ↓
    Train Initial Model
            ↓
 Predict Pseudo Labels for Unlabeled Data
            ↓
    Retrain Model on ALL Data
            ↓
      Improved Model
```

## 4. Key Terminology
*   **Labeled Data:** Known answers (e.g., Image1 → Cancer).
*   **Unlabeled Data:** Unknown answers (e.g., Image3 → ???).
*   **Pseudo Labels:** Labels predicted by the initial model for the unlabeled portion, which are then used as 'ground truth' for retraining.

## 5. Python Example: Pseudo-Labeling Concept
This example demonstrates how we can use a small labeled set to 'teach' the model how to label the rest of our data.

````python
from sklearn.semi_supervised import LabelPropagation
import numpy as np

# 1. Prepare Data
# -1 indicates 'Unlabeled' data in Scikit-Learn's semi-supervised tools
X_mixed = np.array([[2], [4], [3], [5], [8], [9]])
y_mixed = np.array([0, 1, -1, -1, -1, -1]) # Only first two are labeled

# 2. Initialize and Train Model
# LabelPropagation 'spreads' labels from known to unknown points
lp_model = LabelPropagation()
lp_model.fit(X_mixed, y_mixed)

# 3. View Results
predicted_labels = lp_model.transduction_
print("Original Labels ( -1 = Unlabeled ):", y_mixed)
print("Predicted (Pseudo) Labels for all:  ", predicted_labels)

# Comparison
for val, label in zip(X_mixed.flatten(), predicted_labels):
    status = "(Pseudo Label)" if val > 4 else "(Original Label)"
    print(f"Input: {val} -> Assigned Class: {label} {status}")
````

## 6. Advantages & Limitations
| Advantages | Limitations |
| :--- | :--- |
| Requires fewer labeled samples | Wrong pseudo labels can reduce accuracy |
| Reduces manual annotation costs | More complex training pipeline |
| Scales better with big data | Sensitive to the quality of the initial small dataset |

➡ Even labeling a small dataset may not be feasible. Can a model create labels by itself? This leads to **Self-Supervised Learning**.

# Topic 9: Self-Supervised Learning

## 1. Definition
Self-Supervised Learning is a type of Machine Learning where the system learns to supervise itself. It derives labels automatically from the raw data by hiding part of the input and trying to predict it from the remaining parts.

## 2. How is it different?
*   **Supervised:** Human provides labels ($Y$).
*   **Unsupervised:** No labels ($Y$).
*   **Self-Supervised:** The model creates its own labels ($Y$) from the data ($X$).

## 3. Workflow
```text
Raw Data (Unlabeled)
       ↓
Pretext Task (e.g., Hide a word in a sentence)
       ↓
Model predicts the hidden part
       ↓
Model learns features/representations
       ↓
Apply knowledge to real task (Downstream Task)
```

## 4. Examples in the Real World
*   **NLP (BERT/GPT):** "The cat sat on the [MASK]" $\rightarrow$ Model predicts "mat".
*   **Computer Vision:** Rotate an image by 90 degrees $\rightarrow$ Model learns to predict the rotation angle.
*   **Colorization:** Convert a grayscale image to color $\rightarrow$ Model learns about objects to guess their colors.

## 5. Python Example: The Pretext Task (Simulated)
In this example, we simulate a simple 'Pretext Task' where the model must predict a missing feature in a dataset to learn the relationship between variables.

````python
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Raw Unlabeled Data (e.g., measurements of an object)
# Let's say: [Height, Weight, Volume]
raw_data = np.array([
    [10, 20, 200],
    [15, 30, 450],
    [12, 24, 288],
    [20, 40, 800]
])

# 2. Self-Supervision: Hide 'Volume' and make it the Label (Y)
X_self = raw_data[:, :2]  # Height, Weight
y_self = raw_data[:, 2]   # Volume (The model 'invents' this label from data)

# 3. Train on the self-created task
ss_model = LinearRegression()
ss_model.fit(X_self, y_self)

# 4. Use the model to predict 'Volume' for a new unlabeled object
new_object = np.array([[18, 36]])
predicted_volume = ss_model.predict(new_object)

print(f"Self-Supervised Result: Predicted Volume is {predicted_volume[0]}")
print("The model learned the 'Volume = Height * Weight' rule without external labels!")
````

## 6. Advantages & Limitations
| Advantages | Limitations |
| :--- | :--- |
| Can use unlimited unlabeled data (the internet) | Requires clever 'Pretext Tasks' |
| Learns very deep representations | Computationally very heavy |
| Foundation for LLMs (ChatGPT) | Difficult to design for all data types |

# Topic 10: Reinforcement Learning (RL)

## 1. Definition
Reinforcement Learning (RL) is a type of Machine Learning where an **Agent** learns by interacting with an **Environment** and receiving **Rewards** or **Penalties** for its actions. Unlike Supervised Learning, there are no correct answers provided; the agent learns through trial and error.

## 2. Why do we need it?
Some problems cannot provide the correct answer beforehand. For example, when teaching a child to ride a bike, nobody tells them the exact movement for every second. They try, fall, learn, and improve through experience.

## 3. RL Components
```text
            Reward (+/-)
                ▲
                │
Agent ───Action──► Environment
  ▲                  │
  │                  ▼
  └────── State ◄────┘
```

## 4. Important Terminology
*   **Agent:** The learner or decision-maker (Robot, Chess AI).
*   **Environment:** The world the agent operates in (Maze, Road).
*   **State (S):** The current situation (e.g., current board position or speed).
*   **Action (A):** The decision taken by the agent (e.g., turn left, move pawn).
*   **Reward (R):** Feedback indicating how good an action was.
*   **Policy (π):** The strategy followed by the agent (π(a|s)).
*   **Value Function (V):** Expected future reward from a state.
*   **Q-Value (Q):** Expected reward of taking a specific action in a state.

## 5. Exploration vs. Exploitation
*   **Exploration:** Trying new actions to discover better rewards.
*   **Exploitation:** Choosing the best-known action to get a guaranteed reward.

## 6. Formula: Cumulative Reward (Return)
The objective is to maximize the total future reward using a **Discount Factor (γ)**:
$$Return = R_1 + γR_2 + γ^2R_3 + \dots$$
*   **γ = 0:** Only immediate reward matters.
*   **γ ≈ 1:** Future rewards are also important.

## 5. Python Example: The Reward Logic
In this simulation, an agent tries to guess a 'target' number. It receives a positive reward for hitting the target and a negative one for missing, simulating the trial-and-error learning process.

````python
# Simple reward simulation
reward = 0
actions = ["Left", "Right", "Right"]

print("Agent performing sequence of actions...")

for action in actions:
    if action == "Right":
        reward += 10
        print(f"Action: {action} | Reward: +10")
    else:
        reward -= 5
        print(f"Action: {action} | Reward: -5")

print("\nTotal Reward:", reward)
````

## 7. Working Example: Robot Maze
1. **Start** → Move Right (Reward +5)
2. **Move Up** → (Reward +10)
3. **Reach Goal** → (Reward +100)
*   **Wrong move (Hit Wall):** Reward -20.

After many trials, the robot learns the shortest path.

## 8. Advantages & Limitations
| Advantages | Limitations |
| :--- | :--- |
| Learns without labels | Needs many interactions |
| Learns optimal strategies | Training can be slow |
| Continual improvement | Computationally expensive |
