### Topic 2 : Information

#### 1. Definition
Information is processed, organized, and meaningful data that helps in understanding, analysis, or decision-making.

* **Data** is the input.
* **Information** is the output after processing the data.

#### 2. Why do we need it?
Raw data alone cannot help us make decisions. We first convert it into meaningful information.

**Example from our Student Data:**
* **Raw Data:** Student A (2 hrs), Student D (8 hrs).
* **Conclusion/Information:** Students studying more than 6 hours have a higher chance of passing.

#### 3. Formula
Data → Processing → Information → Decision → Knowledge

#### 4. Example
* **Raw Data:** Hours [2, 4, 5, 8]
* **Information:** Average Hours Studied = 4.75, Maximum Hours = 8

#### 5. Working
Raw Data → Cleaning → Analysis → Visualization → Information → Decision

#### 6. Where is it used?
Every Data Science project (Sales Reports, Business Dashboards, Medical Reports, etc.).

#### 7. Advantages
* Easier to understand
* Supports decision making
* Removes ambiguity
* Reveals hidden patterns

#### 8. Limitations
Information is only as good as the data. Poor data leads to poor information and wrong decisions.

#### 9. Python Example
We take our raw list of hours and process it to find the average, turning data into information.

```python
hours = [2, 4, 5, 8]

average = sum(hours) / len(hours)

print("Average Hours =", average)

# Notice:
# [2, 4, 5, 8] (Raw Data)
# ↓
# Average = 4.75 (Useful Information)
```
